```python
comparacion = happiness[happiness['Year'].isin([2019, 2021])]
comparacion_pivot = comparacion.pivot(index ='Country name', columns ='Year', values='Life evaluation (3-year average)')
comparacion_pivot['cambio'] = comparacion_pivot[2021] - comparacion_pivot[2019]
comparacion_pivot.sort_values('cambio').head(10)
```

```python
top_caidas = comparacion_pivot.sort_values('cambio').head(10)

plt.figure(figsize=(10,6))
plt.barh(top_caidas.index, top_caidas['cambio'], color='crimson')
plt.title('Países con mayor caída en felicidad (2019 vs 2021)')
plt.xlabel('Cambio en puntaje de felicidad')
plt.gca().invert_yaxis()
plt.show()
```

```python
top_subidas = comparacion_pivot.sort_values('cambio', ascending=False).head(10)

plt.figure(figsize=(10,6))
plt.barh(top_subidas.index, top_subidas['cambio'], color='seagreen')
plt.title('Países con mayor mejora en felicidad (2019 vs 2021)')
plt.xlabel('Cambio en puntaje de felicidad')
plt.gca().invert_yaxis()
plt.show()
```

```python
happiness['Nivel_GDP'] = pd.cut(
    happiness['Explained by: Log GDP per capita'],
    bins=3,
    labels=['PIB bajo', 'PIB medio', 'PIB alto']
)

plt.figure(figsize=(10,6))
sns.boxplot(
    data=happiness,
    x='Nivel_GDP',
    y='Life evaluation (3-year average)',
    hue='Nivel_GDP',
    palette='Set2',
    legend=False
)
plt.title('Felicidad según nivel de PIB per cápita')
plt.xlabel('Nivel de PIB')
plt.ylabel('Puntaje de felicidad')
plt.show()
```

```python
happiness['Periodo'] = happiness['Year'].apply(lambda x: 'Durante covid (2020-21)' if x in [2020,2021] else 'Otros años')

plt.figure(figsize=(10,6))
sns.boxplot(
    data=happiness,
    x='Nivel_GDP',
    y='Life evaluation (3-year average)',
    hue='Periodo',
    palette='Set1'
)
plt.title('Felicidad por nivel de PIB, comparando covid vs otros años')
plt.xlabel('Nivel de PIB')
plt.ylabel('Puntaje de felicidad')
plt.show()
```

```python
promedio_pais = happiness.groupby('Country name')['Life evaluation (3-year average)'].mean().sort_values(ascending=False)

top3 = promedio_pais.head(3)
bottom3 = promedio_pais.tail(3)
mexico_val = promedio_pais['Mexico']

seleccion = pd.concat([top3, pd.Series({'Mexico': mexico_val}), bottom3])

plt.figure(figsize=(10,6))
colores = ['seagreen']*3 + ['steelblue'] + ['crimson']*3
plt.barh(seleccion.index, seleccion.values, color=colores)
plt.title('Los más felices, México, y los menos felices (promedio histórico)')
plt.xlabel('Puntaje promedio de felicidad')
plt.gca().invert_yaxis()
plt.show()
```

```python
factores = ['Explained by: Log GDP per capita', 'Explained by: Social support', 
            'Explained by: Healthy life expectancy', 'Explained by: Freedom to make life choices']

tendencia_factores = happiness.groupby('Year')[factores].mean()

tendencia_factores.plot(figsize=(10,6), marker='o')
plt.title('Evolución de los factores promedio (2019-2025)')
plt.xlabel('Año')
plt.ylabel('Valor promedio del factor')
plt.legend(title='Factor', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
colores = ['seagreen']*3 + ['steelblue'] + ['crimson']*3
bars = plt.barh(seleccion.index, seleccion.values, color=colores)

for bar in bars:
    ancho = bar.get_width()
    plt.text(ancho + 0.05, bar.get_y() + bar.get_height()/2, 
              f'{ancho:.2f}', va='center', fontsize=10)

plt.title('Los más felices, México, y los menos felices (promedio histórico)')
plt.xlabel('Puntaje promedio de felicidad')
plt.gca().invert_yaxis()
plt.show()
```

```python
paises_interes = list(top3.index) + ['Mexico'] + list(bottom3.index)

factores_clave = ['Explained by: Log GDP per capita', 'Explained by: Social support', 
                   'Explained by: Healthy life expectancy', 'Explained by: Generosity',
                   'Explained by: Perceptions of corruption']

tabla_factores = happiness[happiness['Country name'].isin(paises_interes)].groupby('Country name')[factores_clave].mean()
tabla_factores = tabla_factores.reindex(paises_interes)
tabla_factores
```

```python
tabla_factores.plot(kind='bar', figsize=(12,7), colormap='Set2')
plt.title('Comparación de factores por país')
plt.ylabel('Valor del factor')
plt.xlabel('País')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Factor', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
datos_paises = happiness[happiness['Country name'].isin(paises_interes)]

plt.figure(figsize=(10,6))
sns.lineplot(data=datos_paises, x='Year', y='Explained by: Social support', hue='Country name', marker='o')
plt.title('Evolución del apoyo social por país (2019-2025)')
plt.xlabel('Año')
plt.ylabel('Apoyo social')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
